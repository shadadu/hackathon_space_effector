```python
import matplotlib.pyplot as plt
import numpy as np

# Generate sample data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# Plot both lines on the same axes
plt.plot(x, y1, label='Sine Wave', color='blue')
plt.plot(x, y2, label='Cosine Wave', color='red')

# Add decorations
plt.title('Multiple Lines on One Plot')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.legend()  # Displays the labels assigned above

# Render the graph
```
